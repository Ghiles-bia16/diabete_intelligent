```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("diabetes_prediction_dataset.csv")
```

```python
df
```

```python
df.tail()
```

```python
df.describe()
```

```python
plt.hist(df['age'], bins=30, edgecolor='black', color = 'green')
plt.title("Distribution de l'age")
plt.xlabel('Age')
plt.ylabel('Personnes')
plt.show()
```

```python
sns.countplot(x='gender', data=df[df['gender'] != 'Other'])
plt.title('Distribution des genres')
plt.xlabel('Genre')
plt.ylabel('Personnes')
plt.show()
```

```python
df['gender'].value_counts()
```

```python
plt.hist(df['bmi'], bins=30)
plt.title("Distribution de l'IMC")
plt.xlabel('IMC')
plt.ylabel('Personnes')
plt.show()
```

```python
sns.countplot(x='hypertension', data=df)
plt.title("Distribution du diabète")
plt.xlabel('Hyertension')
plt.ylabel('Personnes')
plt.show()
```

```python
sns.countplot(x='heart_disease', data=df)
plt.title("Distribution des maladies cardiaque")
plt.xlabel('Maladie cardiaque')
plt.ylabel('Personnes')
plt.show()
```

```python
sns.countplot(x='diabetes', data=df)
plt.title("Distribution du diabète")
plt.xlabel('Diabète')
plt.ylabel('Personnes')
plt.show()
```

```python
sns.countplot(x='smoking_history', data=df)
plt.title('Distribution des fumeurs')
plt.xlabel('Fumeurs')
plt.ylabel('Personnes')
plt.show()
```

```python
sns.boxplot(x='diabetes', y='bmi', data=df)
plt.ylabel('IMC')
plt.xlabel('Diabète')
plt.title('IMC vs Diabetes')
plt.show()
```

```python
sns.boxplot(x='diabetes', y='age', data=df)
plt.title('Age et Diabète')
plt.ylabel('Age')
plt.xlabel('Diabete')
plt.show()
```

```python
sns.countplot(x='gender', hue='diabetes', data=df[df['gender'] != 'Other'])
plt.title('Genre et Diabetes')
plt.ylabel('Genre')
plt.xlabel('Diabete')
plt.show()
```

```python
sns.boxplot(x='diabetes', y='blood_glucose_level', data=df)
plt.title('Glycémie et Diabetes')
plt.ylabel('Glycémie')
plt.xlabel('Diabète')
plt.show()
```

```python
sns.scatterplot(x='age', y='bmi', hue='diabetes', data=df)
plt.title('Age et IMC')
plt.ylabel('Age')
plt.xlabel('Diabète')
plt.show()
```
